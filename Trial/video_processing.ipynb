```python
def renaming_labels(yolo_ann_dir):
    for sequence in os.listdir(yolo_ann_dir):
        sequence_path =os.path.join(yolo_ann_dir, sequence)
        if not os.path.isdir(sequence_path):
            continue
        for frame_file in os.listdir(sequence_path):
            frame_path = os.path.join(sequence_path, frame_file)
            new_frame_name = frame_file.replace(' ', '_')
            new_frame_path = os.path.join(sequence_path, new_frame_name)
            if os.path.exists(new_frame_path):
                os.rename(frame_path, new_frame_path)
                print(f"Renamed {frame_path} to {new_frame_path}")
            else:
                print(f"Image file {new_frame_name} does not exist for {frame_file}")
        new_sequence_name = sequence.replace(' ', '_')
        new_sequence_path = os.path.join(yolo_ann_dir, new_sequence_name)
        if os.path.exists(new_sequence_path):
            os.rename(sequence_path, new_sequence_path)

renaming_labels('D:/ML/iitiSoC/PS 3/visdrone/videos/VisDrone2019-VID-train/labels')
renaming_labels('D:/ML/iitiSoC/PS 3/visdrone/videos/VisDrone2019-VID-val/labels')
       
```
